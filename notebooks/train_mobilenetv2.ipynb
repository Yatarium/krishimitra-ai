```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
DATA_DIR = '/content/data/filtered'

def random_erasing(img, p=0.5, max_area=0.15):
    if np.random.rand() > p:
        return img
    h, w = img.shape[0], img.shape[1]
    area = h * w
    target_area = np.random.uniform(0.02, max_area) * area
    aspect_ratio = np.random.uniform(0.3, 3.3)
    erase_h = int(round((target_area * aspect_ratio) ** 0.5))
    erase_w = int(round((target_area / aspect_ratio) ** 0.5))
    if erase_h < h and erase_w < w:
        x = np.random.randint(0, h - erase_h)
        y = np.random.randint(0, w - erase_w)
        img[x:x+erase_h, y:y+erase_w, :] = 0
    return img

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=25,
    width_shift_range=0.15,
    height_shift_range=0.15,
    brightness_range=[0.6, 1.4],
    zoom_range=0.2,
    shear_range=0.1,
    horizontal_flip=True,
    preprocessing_function=random_erasing,
    validation_split=0.2
)
val_datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)

train_gen = train_datagen.flow_from_directory(
    DATA_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', subset='training', shuffle=True
)
val_gen = val_datagen.flow_from_directory(
    DATA_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', subset='validation', shuffle=False
)
```

```python
from sklearn.utils.class_weight import compute_class_weight

class_labels = np.unique(train_gen.classes)
weights = compute_class_weight(class_weight='balanced', classes=class_labels, y=train_gen.classes)
class_weights = dict(zip(class_labels, weights))
print(class_weights)
```

```python
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau

# Unfreeze the last 30% of all layers (base + head combined)
total_layers = len(model.layers)
unfreeze_from = int(total_layers * 0.7)   # keep first 70% frozen, unfreeze last 30%

for i, layer in enumerate(model.layers):
    layer.trainable = i >= unfreeze_from

print(f"Total layers: {total_layers}")
print(f"Frozen: {unfreeze_from} | Trainable: {total_layers - unfreeze_from}")

model.compile(optimizer=Adam(learning_rate=1e-5),
              loss='categorical_crossentropy', metrics=['accuracy'])

CHECKPOINT_DIR = '/content/drive/MyDrive/KrishiMitra/models'

callbacks = [
    ModelCheckpoint(f'{CHECKPOINT_DIR}/finetune_best.keras',
                     save_best_only=True, monitor='val_accuracy', mode='max'),
    ModelCheckpoint(f'{CHECKPOINT_DIR}/finetune_latest.keras',
                     save_best_only=False),  # saves every epoch — protects against timeouts
    EarlyStopping(patience=5, restore_best_weights=True, monitor='val_accuracy'),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3)
]

history_finetune = model.fit(
    train_gen, validation_data=val_gen, epochs=10,
    class_weight=class_weights, callbacks=callbacks
)
```
